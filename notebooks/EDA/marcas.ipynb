```python
mapa_marcas = {
    # L'OR
    "L'OR": "L'OR",
    "L'Or": "L'OR",
    "Lor": "L'OR",

    # Orfeu
    "ORFEU": "Orfeu",
    "Orfeu": "Orfeu",

    # Santa Mônica
    "Santa Monica": "Santa Mônica",
    "Santa Mônica": "Santa Mônica",

    # Bravo Café
    "Bravo Cafe": "Bravo Café",
    "Bravo Café": "Bravo Café",

    # 3 Corações
    "Tres Coracoes": "3 Corações",
    "Três": "3 Corações",
    "Tres": "3 Corações",
    "3 Corações": "3 Corações",

    # Qualitá
    "QUALITA": "Qualitá",
    "Qualitá": "Qualitá",
    "Qualitá Exclusive": "Qualitá",

    # Prima Qualitá
    "Prima Qualita": "Prima Qualitá",
    "Prima Qualitá": "Prima Qualitá",

    # Café do Ponto
    "Café Do Ponto": "Café do Ponto",
    "Café do Ponto": "Café do Ponto",
    "do Ponto": "Café do Ponto",

    # Iguaçu
    "Cafe Iguacu": "Iguaçu",
    "Iguaçu": "Iguaçu",

    # erro
    "pa": None
}
```

```python
mix["Fabricante"] = mix["Fabricante"].replace(mapa_marcas)

# remover linhas inválidas
mix = mix[mix["Fabricante"].notna()].copy()

display(utils.tabela_negrito(mix.head(10)))
```

```python
mix_corrigido = mix.groupby("Fabricante", as_index=False).sum()
display(utils.tabela_negrito(mix_corrigido.head(20)))
```

```python
colunas_lojas = [col for col in mix_corrigido.columns if col != "Fabricante"]

mix_corrigido["Total"] = mix_corrigido[colunas_lojas].sum(axis=1)
mix_corrigido = mix_corrigido.sort_values("Total", ascending=False)

display(utils.tabela_negrito(mix_corrigido.head(20)))
```

```python
preco["Fabricante"] = preco["Fabricante"].replace(mapa_marcas)
preco = preco[preco["Fabricante"].notna()].copy()

display(utils.tabela_negrito(preco.head()))
print(preco.columns.tolist())
```
